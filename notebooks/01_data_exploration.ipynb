unique hashtag counts

```python

# Function to count unique hashtag names
def count_unique_hashtags(hashtags):
    names = [
        item["hashtag_name"]
        for item in hashtags
        if item["hashtag_name"] is not None
    ]

    return len(set(names))


# Calculate unique hashtag counts
videos_clean["unique_hashtag_count"] = (
    videos_clean["hashtags"].apply(count_unique_hashtags)
)

# Select videos with an existing recorded count
known_hashtags = videos_clean[
    videos_clean["hashtag_count"].notna()
]

# Compare unique counts with recorded counts
matching = (
    known_hashtags["unique_hashtag_count"]
    == known_hashtags["hashtag_count"]
).sum()

different = (
    known_hashtags["unique_hashtag_count"]
    != known_hashtags["hashtag_count"]
).sum()

print("Matching unique counts:", matching)
print("Different unique counts:", different)
```

```python

# Select records where unique counts do not match
remaining_mismatches = known_hashtags[
    known_hashtags["unique_hashtag_count"]
    != known_hashtags["hashtag_count"]
].copy()

# Calculate the difference
remaining_mismatches["difference"] = (
    remaining_mismatches["unique_hashtag_count"]
    - remaining_mismatches["hashtag_count"]
)

# Display the first 10 records
display(
    remaining_mismatches[
        [
            "video_id",
            "hashtag_count",
            "unique_hashtag_count",
            "difference"
        ]
    ].head(10)
)

# Count whether calculated values are higher or lower
print(
    "Calculated count higher:",
    (remaining_mismatches["difference"] > 0).sum()
)

print(
    "Calculated count lower:",
    (remaining_mismatches["difference"] < 0).sum()
)
```

Ivestigate decreases to 0

```python

# Check how many decreases end at zero

for column in engagement_columns:

    previous = engagement_clean.groupby("video_id")[
        column
    ].shift(1)

    # Count transitions from a positive value to zero
    drops_to_zero = (
        (previous > 0)
        & (engagement_clean[column] == 0)
    ).sum()

    print(f"{column}: {drops_to_zero} drops to zero")
```

Check Whether Multiple Metrics Drop to Zero Together

```python

# Create a temporary count of metrics dropping to zero
drop_count = pd.Series(
    0,
    index=engagement_clean.index
)

for column in engagement_columns:

    previous = engagement_clean.groupby("video_id")[
        column
    ].shift(1)

    drops_to_zero = (
        (previous > 0)
        & (engagement_clean[column] == 0)
    )

    drop_count += drops_to_zero.astype(int)

# Count observations where multiple metrics dropped to zero
print(
    "At least 2 metrics dropped to zero:",
    (drop_count >= 2).sum()
)

print(
    "At least 3 metrics dropped to zero:",
    (drop_count >= 3).sum()
)

print(
    "At least 5 metrics dropped to zero:",
    (drop_count >= 5).sum()
)

print(
    "All 7 metrics dropped to zero:",
    (drop_count == 7).sum()
)
```

Flag suspicious egagement observations

```python

# Flag observations where at least 3 metrics
# dropped from a positive value to zero

engagement_clean["possible_engagement_reset"] = (
    drop_count >= 3
)

# Verify the number of flagged observations
print(
    "Possible engagement resets:",
    engagement_clean["possible_engagement_reset"].sum()
)

# Count the affected videos
print(
    "Videos affected:",
    engagement_clean.loc[
        engagement_clean["possible_engagement_reset"],
        "video_id"
    ].nunique()
)

# Check how many observations remain
print(
    "Total engagement observations:",
    len(engagement_clean)
)
```

do engagement counts still increase after reset

```python

# Find the final observation date for each video
last_date = engagement_clean.groupby("video_id")[
    "date"
].transform("max")

# Identify flagged observations that occur on
# the last recorded date of their video
reset_flag = engagement_clean["possible_engagement_reset"]

resets_at_end = (
    reset_flag
    & (engagement_clean["date"] == last_date)
)

# Count resets at the end and before the end
print(
    "Resets on final observation:",
    resets_at_end.sum()
)

print(
    "Resets followed by later observations:",
    (reset_flag & (engagement_clean["date"] < last_date)).sum()
)
```

whether views recover after suspicious reset

```python

# Get the next recorded view count for each video
next_views = engagement_clean.groupby("video_id")[
    "play_count"
].shift(-1)

# Select suspicious reset observations
reset_flag = engagement_clean["possible_engagement_reset"]

# Check whether views become positive afterward
views_recovered = (
    reset_flag
    & next_views.notna()
    & (next_views > 0)
)

# Check whether views remain zero
views_still_zero = (
    reset_flag
    & next_views.notna()
    & (next_views == 0)
)

# Check whether there is no next observation
no_next_observation = (
    reset_flag
    & next_views.isna()
)

print("Views positive at next observation:", views_recovered.sum())

print("Views still zero at next observation:", views_still_zero.sum())

print("No next observation:", no_next_observation.sum())
```

check whether the views returned to their previous level

```python

# Get views before and after each observation
previous_views = engagement_clean.groupby("video_id")[
    "play_count"
].shift(1)

next_views = engagement_clean.groupby("video_id")[
    "play_count"
].shift(-1)

# Select flagged observations with a previous
# and next recorded view count
valid_resets = (
    engagement_clean["possible_engagement_reset"]
    & previous_views.notna()
    & next_views.notna()
)

# Check whether views recovered to at least
# their previous level
full_recovery = (
    valid_resets
    & (next_views >= previous_views)
)

# Check whether views recovered only partially
partial_recovery = (
    valid_resets
    & (next_views < previous_views)
    & (next_views > 0)
)

print(
    "Views recovered to previous level or higher:",
    full_recovery.sum()
)

print(
    "Views recovered partially:",
    partial_recovery.sum()
)
```

```python

# Exclude flagged observations from the summary calculation
reliable_engagement = engagement_clean.loc[
    ~engagement_clean["possible_engagement_reset"]
]

# Get the last unflagged observation for each video
last_reliable = reliable_engagement.groupby(
    "video_id"
).last()

# Add the last reliable views to our summary
video_engagement_summary["last_reliable_views"] = (
    last_reliable["play_count"]
)

# Add the day of the last reliable observation
video_engagement_summary["last_reliable_day"] = (
    last_reliable["days_since_post"]
)

# Check how many videos have different final values
different_final_views = (
    video_engagement_summary["last_views"]
    != video_engagement_summary["last_reliable_views"]
).sum()

print(
    "Videos with different final view counts:",
    different_final_views
)

print(
    "Missing last reliable views:",
    video_engagement_summary["last_reliable_views"].isna().sum()
)
```

```python

# Calculate the number of days between the first
# observation and the last unflagged observation
video_engagement_summary["reliable_tracking_duration"] = (
    video_engagement_summary["last_reliable_day"]
    - video_engagement_summary["first_day"]
)

# Calculate average daily view growth
video_engagement_summary["avg_daily_view_growth"] = (
    (
        video_engagement_summary["last_reliable_views"]
        - video_engagement_summary["first_views"]
    )
    / video_engagement_summary[
        "reliable_tracking_duration"
    ].replace(0, np.nan)
)

# Display summary statistics
display(
    video_engagement_summary[
        "avg_daily_view_growth"
    ].describe()
)

# Check missing growth values
print(
    "Missing growth values:",
    video_engagement_summary[
        "avg_daily_view_growth"
    ].isna().sum()
)

# Check negative growth values
print(
    "Negative growth values:",
    (
        video_engagement_summary[
            "avg_daily_view_growth"
        ] < 0
    ).sum()
)
```
